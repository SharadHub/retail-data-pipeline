## Business Rules for Handling Nulls
- order_approved_at is null: If approval time is missing, the order was likely abandoned at checkout or cancelled before payment processing, so these rows should be kept for drop-off analysis but filtered out when calculating financial revenue.

- order_delivered_carrier_date is null: If the carrier date is missing, the order has not left the warehouse yet, meaning it should be treated as unfulfilled or backordered rather than shipped.

- order_delivered_customer_date is null: If the customer delivery date is missing, the order is still in transit or cancelled, so these rows should be excluded only when calculating final delivery performance metrics.
